```python

import pandas as pd

# Sample data based on user description
datadf = df.copy()

df_sec = df_sec.copy()

# Function to check if a code falls in a section range or matches single code
def match_section(code_grouper, section):
    if '-' in section:
        start, end = section.split('-')
        # Compare alphanumeric codes lexicographically
        return start <= code_grouper <= end
    else:
        return code_grouper == section

# For each row in datadf, find matching section
section_list = []
section_name_list = []

for cg in datadf['Code_Grouper']:
    matched_section = None
    matched_name = None
    for sec, name in zip(df_sec['Section'], df_sec['Section_Name']):
        if match_section(cg, sec):
            matched_section = sec
            matched_name = name
            break
    section_list.append(matched_section)
    section_name_list.append(matched_name)

# Add to datadf
datadf['Section'] = section_list
datadf['Section_Name'] = section_name_list

print(datadf)
```

```python
import pandas as pd

# ----- Step 1: Clean Section range -----
datadf[['sec_start', 'sec_end']] = datadf['Section'].str.split('-', expand=True)
datadf['sec_start'] = datadf['sec_start'].str.strip()
datadf['sec_end'] = datadf['sec_end'].str.strip()

# ----- Step 2: Extract chapter ranges -----
df_chap[['chapter_label', 'chapter_range']] = df_chap['Chapter'].str.split(' ', n=1, expand=True)
df_chap[['chap_start', 'chap_end']] = df_chap['chapter_range'].str.split('-', expand=True)
df_chap['chap_start'] = df_chap['chap_start'].str.strip()
df_chap['chap_end']   = df_chap['chap_end'].str.strip()

# ----- Helper: safely check if code is between start and end -----
def belongs_to_chapter(code, chap_start, chap_end):
    if pd.isna(code) or pd.isna(chap_start) or pd.isna(chap_end):
        return False
    return chap_start <= code <= chap_end  # works for ICD codes as strings

# ----- Step 3: Map chapter using sec_start -----
def map_chapter(row):
    code = row['sec_start']
    for _, chap in df_chap.iterrows():
        if belongs_to_chapter(code, chap['chap_start'], chap['chap_end']):
            return pd.Series([chap['Chapter'], chap['Chapter_Name']])
    return pd.Series([None, None])

datadf[['Chapter', 'Chapter_Name']] = datadf.apply(map_chapter, axis=1)

print(datadf)

```
